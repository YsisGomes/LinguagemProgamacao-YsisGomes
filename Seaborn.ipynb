# Atividade Prática: Visualização Estatística de Dados com Seaborn

## Contexto do Problema

Atuação como Analista de Dados em uma empresa de software (SaaS), com uma base contendo o registro de interação dos usuários com o sistema. O objetivo é compreender o perfil de engajamento, identificar padrões de uso entre diferentes planos de assinatura e visualizar a correlação entre problemas reportados e a satisfação do cliente.

## 1. Configuração do Ambiente e Base de Dados

Importação das bibliotecas e geração do DataFrame sintético da atividade.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid", palette="muted")
```

### Geração dos dados sintéticos

Criação de 300 registros de interação com variáveis de idade, plano, tempo de uso, bugs reportados e satisfação.

```python
np.random.seed(101)
n = 300

dados = {
    'idade': np.random.normal(35, 10, n).astype(int),
    'plano': np.random.choice(['Gratuito', 'Básico', 'Pro'], n, p=[0.5, 0.3, 0.2]),
    'tempo_uso_horas': np.random.uniform(1, 50, n),
    'bugs_reportados': np.random.poisson(2, n),
    'satisfacao': np.random.randint(1, 11, n)
}

df_sistema = pd.DataFrame(dados)
```

### Ajuste das regras de negócio sintéticas

Aplicação de regras para gerar correlações visuais coerentes entre plano, tempo de uso, satisfação e bugs reportados.

```python
df_sistema.loc[df_sistema['plano'] == 'Pro', 'tempo_uso_horas'] += 15
df_sistema.loc[df_sistema['plano'] == 'Pro', 'satisfacao'] += 2
df_sistema['satisfacao'] = df_sistema['satisfacao'] - (df_sistema['bugs_reportados'] * 0.5)
df_sistema['satisfacao'] = df_sistema['satisfacao'].clip(1, 10).astype(int)
df_sistema['idade'] = df_sistema['idade'].clip(18, 70)

df_sistema.head()
```

## 2. Roteiro de Tarefas

## Parte 1: Distribuições e Contagens (Básico)

### Tarefa 1: Contagem de usuários por plano

Gráfico de barras com `countplot` para exibir a contagem de usuários por tipo de plano, ordenando as barras pela frequência.

```python
sns.countplot(data=df_sistema, x='plano',
              order=df_sistema['plano'].value_counts().index)
plt.title('Contagem de usuários por plano')
plt.show()
```

### Tarefa 2: Distribuição da idade

Análise da distribuição da variável `idade` com `histplot`, adicionando a curva de densidade (KDE).

```python
sns.histplot(data=df_sistema, x='idade', kde=True)
plt.title('Distribuição da idade dos usuários')
plt.show()
```

## Parte 2: Relações Categóricas e Numéricas (Intermediário)

### Tarefa 3: Boxplot do tempo de uso por plano

Comparação da distribuição de `tempo_uso_horas` entre os diferentes tipos de plano.

```python
sns.boxplot(data=df_sistema, x='plano', y='tempo_uso_horas')
plt.title('Tempo de uso por plano')
plt.show()
```

### Tarefa 4: Violinplot de plano vs satisfação

Cruzamento do `plano` com a `satisfacao`, mostrando não apenas a mediana mas a densidade das notas dadas por cada plano.

```python
sns.violinplot(data=df_sistema, x='plano', y='satisfacao')
plt.title('Satisfação por plano')
plt.show()
```

## Parte 3: Correlações e Análise Multivariada (Avançado)

### Tarefa 5: Scatterplot multivariado

Relação entre `tempo_uso_horas` (X) e `satisfacao` (Y), com `hue` por plano e `size` variando conforme `bugs_reportados`.

```python
sns.scatterplot(data=df_sistema, x='tempo_uso_horas', y='satisfacao',
                hue='plano', size='bugs_reportados')
plt.title('Tempo de uso vs satisfação')
plt.show()
```

### Tarefa 6: Matriz de correlação com heatmap

Cálculo da correlação de Pearson apenas para variáveis numéricas e visualização com `heatmap`, paleta divergente e anotações numéricas.

```python
numericas = df_sistema.select_dtypes(include='number')
correlacao = numericas.corr(method='pearson')

sns.heatmap(correlacao, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Matriz de correlação de Pearson')
plt.show()
```

### Tarefa 7 (Bônus): Pairplot

Geração de um `pairplot` envolvendo `tempo_uso_horas`, `bugs_reportados` e `satisfacao`, com o plano como `hue`.

```python
sns.pairplot(df_sistema,
             vars=['tempo_uso_horas', 'bugs_reportados', 'satisfacao'],
             hue='plano')
plt.show()
```
