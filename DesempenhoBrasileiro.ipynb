### Exercício: Análise de Desempenho de Clubes no Campeonato Brasileiro 2024

**Objetivo**
Os alunos irão aprender a:
- Manipular dados de um campeonato de futebol.
- Calcular e visualizar estatísticas de desempenho dos clubes.

### Passo 1: Leitura da Planilha
Carregar os dados do arquivo `brasileirao.xlsx` usando pandas.

```python
import pandas as pd

df = pd.read_excel('brasileirao.xlsx')

df.head()
```

### Passo 2: Calculas o saldo de gols(SG)

Adicionar uma nova coluna chamada "Saldo de Gols" que será a diferença entre Gols Pró (GP) e Gols Contra (GC).

```python
df['Saldo de Gols'] = df['GP'] - df['GC']
df.head()
```

### Passo 3: Calcular a pontuação total

Criar uma coluna chamada "Pontos" onde a pontuação será calculada da seguinte forma:
- Vitória (V) = 3 pontos
- Empate (E) = 1 ponto
- Derrota (D) = 0 pontos

```python
df['Pontos'] = df['V'] * 3 + df['E']
df.head(20)
```

### Passo 4: Filtrar os clubes com mais de 40 pontos

Filtrar os clubes que tem mais de 40 pontos

```python
clubes_maior_40_pontos = df[df['Pontos'] > 40]
clubes_maior_40_pontos
```

### Passo 5: Visualizar os Dados

Criar um gráfico de barras para visualizar a pontuação dos clubes

```python
df.plot(x='Clube', y='Pontos', kind='bar', title='Pontuação dos clubes no Brasil')
```
